```python
# Incremental Merkle tree

ZERO = 0
# 0 = Leaf nodes
HEIGHT = 3

def hash(a, b):
    return f'h({a},{b})'

class MerkleTree:
    def __init__(self):
        # Calculate tree with all leaves = ZERO
        zeros = [ZERO] * HEIGHT
        for i in range(HEIGHT - 1):
            zeros[i + 1] = hash(zeros[i], zeros[i])

        # Zero hash at each level
        self.zeros = zeros
        # Right most left node, where all descendents are filled, at each level
        self.nodes = [ZERO] * HEIGHT
        # Next leaf index
        self.next = 0
        # Leaves
        self.vals = [ZERO] * (2**HEIGHT)

    def root(self):
        h = self.zeros[0]
        k = self.next
        
        for i in range(HEIGHT):
            if k % 2 == 1:
                h = hash(self.nodes[i], h)
            else:
                h = hash(h, self.zeros[i])
            k //= 2

        return h

    def insert(self, val):
        assert self.next < 2 ** HEIGHT - 1

        self.vals[self.next] = val
        self.next += 1

        h = val
        k = self.next

        for i in range(HEIGHT):
            if k % 2 == 1:
                self.nodes[i] = h
                return
            else:
                h = hash(self.nodes[i], h)
            k //= 2
```

```python
tree = MerkleTree()

print(0, tree.root())

for i in range(1, 2**HEIGHT):
    tree.insert(i)
    print(i, tree.root())

print(tree.vals)
```
